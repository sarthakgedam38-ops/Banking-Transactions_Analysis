```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("indian_banking_transactions.csv")
print(df.head())
```

```python
print(df.info())
print(df.describe)
print(df.isnull().sum())
print(df.dtypes)
print(df.duplicated().sum())
```

```python
df['loan_type'] = df['loan_type'].fillna('None')
```

```python
df['transaction_date'] = pd.to_datetime(df['transaction_date'])
print(df.isnull().sum())
print(df['transaction_date'])
print(df.dtypes)
```

```python
df['year'] = df['transaction_date'].dt.year
df['month_name'] = df['transaction_date'].dt.month_name()

df.columns = df.columns.str.title()

print(df.isnull().sum())
```

```python
df.to_csv("Clean_Banking_Transactions.csv")
```
